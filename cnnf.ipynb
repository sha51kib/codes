```python
import tensorflow as tf
from tensorflow.keras import layers, models

# 1. Folder paths
train_path = r"C:\Users\turzo\Documents\Neural\Dataset\Train"
val_path   = r"C:\Users\turzo\Documents\Neural\Dataset\Validation"
test_path  = r"C:\Users\turzo\Documents\Neural\Dataset\Test"

# 2. Load datasets
train_data = tf.keras.utils.image_dataset_from_directory(
    train_path,
    image_size=(128, 128),
    batch_size=32
)

val_data = tf.keras.utils.image_dataset_from_directory(
    val_path,
    image_size=(128, 128),
    batch_size=32
)

test_data = tf.keras.utils.image_dataset_from_directory(
    test_path,
    image_size=(128, 128),
    batch_size=32
)

print("Classes:", train_data.class_names)

# 3. Simple CNN model
model = models.Sequential([
    layers.Rescaling(1./255, input_shape=(128, 128, 3)),

    layers.Conv2D(16, 3, activation="relu"),
    layers.MaxPooling2D(),

    layers.Conv2D(32, 3, activation="relu"),
    layers.MaxPooling2D(),

    layers.Flatten(),
    layers.Dense(64, activation="relu"),
    layers.Dense(3, activation="softmax")
])

# 4. Compile model
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# 5. Train model
model.fit(
    train_data,
    validation_data=val_data,
    epochs=5
)

# 6. Test model
loss, accuracy = model.evaluate(test_data)

print("Test Accuracy:", accuracy)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array

# যে image test করবে তার path
img_path = r"C:\Users\turzo\Documents\Neural\Dataset\Test\Birds\090.jpg"

# Image load and resize
img = load_img(img_path, target_size=(128, 128))

# Image to array
img_array = img_to_array(img)

# Batch dimension add: (128,128,3) -> (1,128,128,3)
img_array = np.expand_dims(img_array, axis=0)

# Prediction
prediction = model.predict(img_array)

# Predicted class
class_names = train_data.class_names
predicted_class = class_names[np.argmax(prediction)]

print("Predicted Class:", predicted_class)

# Show image
plt.imshow(img)
plt.title("Predicted: " + predicted_class)
plt.axis("off")
plt.show()
```
